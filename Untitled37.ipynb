```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import numpy as np

# Function to perform ANOVA and Tukey's HSD post-hoc test
def perform_anova_and_tukey(df, variable, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable} in {location}")

    # Reshape data for ANOVA: create a 'Treatment' column and a 'Value' column
    data_melted = df.melt(value_vars=[col for col in df.columns if variable in col and ('AG_' in col or 'AV_' in col or 'PV_' in col or 'WS' in col)],
                           var_name='Treatment', value_name='Value')

    # Extract treatment type from column names
    def get_treatment(col_name):
        if 'AG_' in col_name:
            return 'AG'
        elif 'AV_' in col_name:
            return 'AV'
        elif 'PV_' in col_name:
            return 'PV'
        elif 'WS' in col_name:
            return 'WS'
        else:
            return 'Other' # Should not happen with correct filtering

    data_melted['Treatment'] = data_melted['Treatment'].apply(get_treatment)

    # Drop rows with NaN values in 'Value' (although imputation should have handled this)
    data_melted = data_melted.dropna(subset=['Value'])

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable} in {location}: Not enough unique treatments.")
        return

    try:
        # Perform ANOVA
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        print("ANOVA Results:")
        display(anova_table)

        # Perform Tukey's HSD post-hoc test
        # Ensure there are at least two groups to compare
        if data_melted['Treatment'].nunique() >= 2:
             tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
             print("\nTukey's HSD Post-Hoc Test Results:")
             display(tukey_result)
        else:
             print("\nSkipping Tukey's HSD: Not enough unique groups for comparison.")


    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable} in {location}: {e}")


# Identify key variables for analysis based on prompt
# Assuming key variables are Soil Moisture, Soil Temp, Ambient Humidity, Ambient Temp, PV Radiation, PV Panel Temp, Precipitation
key_variables = ['Soil_Moisture', 'Soil_Temp', 'Ambient_Humidity', 'Ambient_Temp', 'PV_Radiation', 'PV_Panel_Temp', 'Precipitation']


# Perform analysis for Tamale
for var in key_variables:
     # Filter columns to include only the current variable across relevant treatments and crops
     relevant_cols_tamale = [col for col in tamale_df.columns if var in col and any(treat in col for treat in ['AG_', 'AV_', '_PV', '_WS'])]
     if relevant_cols_tamale:
          perform_anova_and_tukey(tamale_df[relevant_cols_tamale], var, 'Tamale')
     else:
          print(f"\nNo relevant columns found for {var} in Tamale.")


# Perform analysis for Kumasi
for var in key_variables:
    # Filter columns to include only the current variable across relevant treatments and crops
    relevant_cols_kumasi = [col for col in kumasi_df.columns if var in col and any(treat in col for treat in ['AG_', 'AV_', '_PV', '_WS'])]
    if relevant_cols_kumasi:
        perform_anova_and_tukey(kumasi_df[relevant_cols_kumasi], var, 'Kumasi')
    else:
         print(f"\nNo relevant columns found for {var} in Kumasi.")
```

**Reasoning**:
The previous attempt at comparative analysis failed because the column filtering did not correctly group columns by variable and treatment for comparison. The current column naming convention includes both treatment and crop in many names. I need to adjust the approach to perform ANOVA and Tukey's HSD for each variable and *crop type* combination, comparing the 'AG' and 'AV' treatments. For variables without crop types (Ambient, PV, Precipitation), I will compare treatments 'WS', 'PV', 'AV' and 'AG' as relevant.



```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import numpy as np

# Function to perform ANOVA and Tukey's HSD post-hoc test for specific treatments
def perform_anova_and_tukey_treatments(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable} ({crop}) in {location}")

    # Select columns for the specific variable and crop for AG and AV treatments
    ag_col = f'{variable} AG_{crop}' if crop else f'{variable} AG'
    av_col = f'{variable} AV_{crop}' if crop else f'{variable} AV'
    pv_col = f'{variable}_PV' if variable in ['PV_Radiation', 'PV_Panel_Temp'] else None
    ws_col = f'{variable}_WS' if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation'] else None

    cols_to_compare = []
    if ag_col in df.columns:
        cols_to_compare.append(ag_col)
    if av_col in df.columns:
        cols_to_compare.append(av_col)
    if pv_col and pv_col in df.columns:
        cols_to_compare.append(pv_col)
    if ws_col and ws_col in df.columns:
        cols_to_compare.append(ws_col)

    if len(cols_to_compare) < 2:
        print(f"Skipping {variable} ({crop}) in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return

    # Create a melted DataFrame for ANOVA
    data_melted = df[cols_to_compare].melt(var_name='Treatment', value_name='Value')

    # Extract simplified treatment names
    def simplify_treatment_name(col_name, var, crop):
        if f'{variable} AG_{crop}' in col_name:
            return 'AG'
        elif f'{variable} AV_{crop}' in col_name:
            return 'AV'
        elif f'{variable}_PV' in col_name:
            return 'PV'
        elif f'{variable}_WS' in col_name:
             return 'WS'
        else:
            return 'Other'

    if crop:
         data_melted['Treatment'] = data_melted['Treatment'].apply(lambda x: simplify_treatment_name(x, variable, crop))
    else:
         data_melted['Treatment'] = data_melted['Treatment'].replace({f'{variable}_PV': 'PV', f'{variable}_WS': 'WS'})


    # Drop any remaining NaN values
    data_melted = data_melted.dropna(subset=['Value'])

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable} ({crop}) in {location}: Not enough unique treatments ({data_melted['Treatment'].nunique()}).")
        return

    try:
        # Perform ANOVA
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        print("ANOVA Results:")
        display(anova_table)

        # Perform Tukey's HSD post-hoc test
        if data_melted['Treatment'].nunique() >= 2:
            tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
            print("\nTukey's HSD Post-Hoc Test Results:")
            display(tukey_result)
        else:
            print("\nSkipping Tukey's HSD: Not enough unique groups for comparison.")


    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable} ({crop}) in {location}: {e}")


# Define variables and crops
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy', 'PV_Radiation', 'PV_PanelTemp']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']


# Perform analysis for Tamale
print("--- Tamale Analysis ---")
for var in variables_with_crops:
    for crop in crops:
        # Handle the slightly different naming for PV Radiation and Panel Temp within AV plot
        if var in ['PV_Radiation', 'PV_PanelTemp'] and 'AV_' in var:
             perform_anova_and_tukey_treatments(tamale_df, var.replace('AV_', ''), crop, 'Tamale') # Remove AV_ for function
        else:
            perform_anova_and_tukey_treatments(tamale_df, var, crop, 'Tamale')

for var in ambient_variables:
     perform_anova_and_tukey_treatments(tamale_df, var, None, 'Tamale') # None for crop

# Perform analysis for Kumasi
print("\n--- Kumasi Analysis ---")
for var in variables_with_crops:
    for crop in crops:
         # Handle the slightly different naming for PV Radiation and Panel Temp within AV plot
         if var in ['PV_Radiation', 'PV_PanelTemp'] and 'AV_' in var:
              perform_anova_and_tukey_treatments(kumasi_df, var.replace('AV_', ''), crop, 'Kumasi') # Remove AV_ for function
         else:
             perform_anova_and_tukey_treatments(kumasi_df, var, crop, 'Kumasi')

for var in ambient_variables:
     perform_anova_and_tukey_treatments(kumasi_df, var, None, 'Kumasi') # None for crop
```

```python
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pandas as pd
import numpy as np
import re
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd


# Create directories to save outputs
output_dir_tamale = 'tamale_analysis'
output_dir_kumasi = 'kumasi_analysis'
os.makedirs(output_dir_tamale, exist_ok=True)
os.makedirs(output_dir_kumasi, exist_ok=True)


# Function to perform ANOVA and Tukey's HSD and store results (reusing the refined function)
def perform_anova_and_tukey_store_results(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}")

    cols_to_compare = []
    if crop:
        if 'Soil_' in variable:
             ag_pattern = f'{variable} {crop}'
             av_pattern = f'{variable} AV {crop}'
        elif 'plant canopy' in variable:
             ag_pattern = f'{variable}_AG_{crop}'
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}'
        else:
             ag_pattern = f'{variable}.*AG_{crop}'
             av_pattern = f'{variable}.*AV_{crop}'

        ag_cols = [col for col in df.columns if re.search(ag_pattern, col, re.IGNORECASE)]
        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        cols_to_compare.extend(ag_cols)
        cols_to_compare.extend(av_cols)

    else:
        if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']:
            base_pattern = f'^{variable}$'
            base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
            cols_to_compare.extend(base_cols)
        elif variable in ['PV_Radiation', 'PV_Panel_Temp']:
             base_pattern = f'^{variable}$'
             av_patterns = [f'{variable}_AV_{c}' for c in ['tomato', 'pepper', 'eggplant']] + [f'{variable}_Agrivoltaics plot {c}' for c in ['tomato', 'pepper', 'eggplant']]

             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)


    cols_to_compare = list(set(cols_to_compare))

    if len(cols_to_compare) < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return None, None, None # Return None for tables


    filtered_df = df[cols_to_compare]
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])

    def simplify_treatment_name(col_name):
        if 'AG_' in col_name or ('Soil_Moisture ' in col_name and 'AV' not in col_name) or ('Soil_Temp ' in col_name and 'AV' not in col_name):
            return 'AG'
        elif 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_Panel_Temp' in col_name) and not ('AV' in col_name or 'Agrivoltaics plot' in col_name):
            return 'PV'
        elif '_WS' in col_name or 'Ambient_Humidity' in col_name or 'Ambient_Temp' in col_name or 'Precipitation' in col_name:
            return 'WS'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough unique treatments after dropping NaNs ({data_melted['Treatment'].nunique()}).")
        return data_melted, None, None # Return data_melted for plotting even if ANOVA skipped

    try:
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)

        if data_melted['Treatment'].nunique() >= 2:
             tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
             tukey_table = pd.DataFrame(data=tukey_result._results_table.data[1:], columns=tukey_result._results_table.data[0])
        else:
             tukey_result = None
             tukey_table = None


        return data_melted, anova_table, tukey_table

    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: {e}")
        return data_melted, None, None # Return data_melted for plotting even if ANOVA failed


# Function to create box plots using Plotly Express
def create_plotly_box_plot(data, variable, crop, location, output_dir, unit=''):
    if data is None or data.empty:
        print(f"Skipping Plotly box plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No data available.")
        return

    fig = px.box(data,
                 x="Treatment",
                 y="Value",
                 points="all",
                 title=f'{variable} ({crop if crop else "Ambient/PV"}) in {location} by Treatment',
                 labels={'Treatment': 'Treatment', 'Value': f'{variable}{f" ({unit})" if unit else ""}'},
                 color="Treatment", # Color by treatment
                 color_discrete_map={'AG': 'green', 'AV': 'blue', 'PV': 'gray', 'WS': 'orange'}) # Swapped green and blue


    # Save plot as HTML
    plot_filename_html = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_boxplot.html'
    fig.write_html(plot_filename_html)
    print(f"Saved Plotly box plot to {plot_filename_html}")


# Function to create bar plots with error bars (using descriptive stats) - Removed due to persistent errors
# def create_bar_plot_with_error_bars(data, variable, crop, location, output_dir, unit=''):
#      pass # Placeholder for the removed function


# Function to generate descriptive statistics table
def generate_descriptive_stats_table(df, variables, crops, ambient_vars, pv_vars, location, output_dir):
    print(f"\nGenerating descriptive statistics table for {location}")
    all_descriptive_stats = {}

    # Soil and Plant Canopy variables with crops
    for var in variables:
        for crop in crops:
            cols = []
            if 'Soil_' in var:
                cols.extend([f'{var} {crop}', f'{var} AV {crop}'])
            elif 'plant canopy' in var:
                cols.extend([f'{var}_AG_{crop}', f'{var}_Agrivoltaic plot for {crop}'])

            existing_cols = [col for col in cols if col in df.columns]
            if existing_cols:
                temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
                if not temp_df.empty:
                     # Simplify treatment names for the table
                     temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AG' if 'AG' in col_name or ('Soil_' in col_name and 'AV' not in col_name) else 'AV')
                     descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                     all_descriptive_stats[f'{var}_{crop}'] = descriptive_stats


    # Ambient variables
    for var in ambient_vars:
        cols = [var]
        existing_cols = [col for col in cols if col in df.columns]
        if existing_cols:
             temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
             if not temp_df.empty:
                 temp_df['Treatment'] = 'WS' # Assign WS treatment for ambient variables
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    # PV variables
    for var in pv_vars:
         cols = [var] + [f'{var}_AV_{c}' for c in crops] + [f'{var}_Agrivoltaics plot {c}' for c in crops]
         existing_cols = [col for col in cols if col in df.columns]
         if existing_cols:
              temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
              if not temp_df.empty:
                 # Simplify treatment names for PV variables
                 temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AV' if 'AV' in col_name or 'Agrivoltaics plot' in col_name else 'PV')
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    if all_descriptive_stats:
        # Combine all descriptive stats into a single DataFrame for easier saving
        combined_descriptive_stats = pd.concat(all_descriptive_stats, axis=1)
        table_filename = f'{output_dir}/descriptive_statistics.csv'
        combined_descriptive_stats.to_csv(table_filename)
        print(f"Saved descriptive statistics table to {table_filename}")
        display(combined_descriptive_stats.head()) # Display head as table can be large
    else:
        print(f"No descriptive statistics generated for {location}.")


# Store ANOVA and Tukey results
anova_results = {}
tukey_results = {}

# Define variables and crops
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp']


# Perform analysis and generate visualizations/tables for Tamale
print("--- Tamale Analysis Outputs ---")
generate_descriptive_stats_table(tamale_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Tamale', output_dir_tamale)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(tamale_df, var, crop, 'Tamale')
        if data_melted is not None:
            # Create visualizations for AG vs AV
            create_plotly_box_plot(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            # create_bar_plot_with_error_bars(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '') # Removed

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Tamale_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Tamale_{var}_{crop}'] = tukey_table

# Perform analysis and generate visualizations/tables for Kumasi
print("\n--- Kumasi Analysis Outputs ---")
generate_descriptive_stats_table(kumasi_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Kumasi', output_dir_kumasi)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(kumasi_df, var, crop, 'Kumasi')
        if data_melted is not None:
             # Create visualizations for AG vs AV
            create_plotly_box_plot(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            # create_bar_plot_with_error_bars(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '') # Removed

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Kumasi_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Kumasi_{var}_{crop}'] = tukey_table

# Save ANOVA results to Excel
if anova_results:
    anova_output_filename = 'anova_results.xlsx'
    with pd.ExcelWriter(anova_output_filename) as writer:
        for key, table in anova_results.items():
            table.to_excel(writer, sheet_name=key)
    print(f"\nSaved ANOVA results to {anova_output_filename}")

# Save Tukey HSD results to Excel
if tukey_results:
    tukey_output_filename = 'tukey_hsd_results.xlsx'
    with pd.ExcelWriter(tukey_output_filename) as writer:
        for key, table in tukey_results.items():
             table.to_excel(writer, sheet_name=key)
    print(f"Saved Tukey HSD results to {tukey_output_filename}")
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import numpy as np
import re

# Function to perform ANOVA and Tukey's HSD post-hoc test for specific treatments
def perform_anova_and_tukey_treatments_retry(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}")

    # Dynamically find columns based on variable and crop patterns
    cols_to_compare = []
    if crop:
        # For variables with crops, look for AG_ and AV_ treatments
        # Handle Soil variables
        if 'Soil_' in variable:
             ag_pattern = f'{variable} {crop}'
             av_pattern = f'{variable} AV {crop}'
        # Handle Plant canopy variables
        elif 'plant canopy' in variable:
             ag_pattern = f'{variable}_AG_{crop}'
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}'
        else:
             ag_pattern = f'{variable}.*AG_{crop}'
             av_pattern = f'{variable}.*AV_{crop}'


        # Find columns matching the patterns
        ag_cols = [col for col in df.columns if re.search(ag_pattern, col, re.IGNORECASE)]
        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        cols_to_compare.extend(ag_cols)
        cols_to_compare.extend(av_cols)

    else:
        # For ambient/PV variables, look for relevant columns.
        if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']:
            # These variables appear to have only one column representing the WS treatment, so skip ANOVA
            print(f"Skipping {variable} in {location}: Only one treatment (WS) available.")
            return
        elif variable in ['PV_Radiation', 'PV_Panel_Temp']:
             # Look for base PV, and AV specific PV columns for each crop
             base_pattern = f'^{variable}$'
             # Collect all AV-related PV columns regardless of crop when crop is None
             av_patterns = [f'{variable}_AV_{c}' for c in crops] + [f'{variable}_Agrivoltaics plot {c}' for c in crops]

             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)


    # Ensure unique columns in case of overlapping patterns
    cols_to_compare = list(set(cols_to_compare))


    if len(cols_to_compare) < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return

    # Filter the DataFrame to include only the identified column names
    filtered_df = df[cols_to_compare]

    # Melt the filtered DataFrame for ANOVA
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value')

    # Extract simplified treatment names ('AG', 'AV', 'PV', 'WS')
    def simplify_treatment_name(col_name):
        if 'AG_' in col_name or ('Soil_Moisture ' in col_name and 'AV' not in col_name) or ('Soil_Temp ' in col_name and 'AV' not in col_name):
            return 'AG'
        elif 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_Panel_Temp' in col_name) and not ('AV' in col_name or 'Agrivoltaics plot' in col_name):
            # This condition tries to capture the base PV columns as 'PV'
            return 'PV'
        elif '_WS' in col_name or 'Ambient_Humidity' in col_name or 'Ambient_Temp' in col_name or 'Precipitation' in col_name:
            return 'WS'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)


    # Drop any rows with NaN values in the 'Value' column
    data_melted = data_melted.dropna(subset=['Value'])

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough unique treatments after dropping NaNs ({data_melted['Treatment'].nunique()}).")
        return

    try:
        # Perform ANOVA
        print("ANOVA Results:")
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        display(anova_table)

        # Perform Tukey's HSD post-hoc test
        if data_melted['Treatment'].nunique() >= 2:
            print("\nTukey's HSD Post-Hoc Test Results:")
            tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
            display(tukey_result)
        else:
            print("\nSkipping Tukey's HSD: Not enough unique groups for comparison after dropping NaNs.")

    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: {e}")

# Define variables and crops based on observed column names
# Note the long names for ambient/canopy variables
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp']


# Perform analysis for Tamale
print("--- Tamale Analysis (Retry) ---")
for var in variables_with_crops:
    for crop in crops:
        perform_anova_and_tukey_treatments_retry(tamale_df, var, crop, 'Tamale')

for var in ambient_variables:
    perform_anova_and_tukey_treatments_retry(tamale_df, var, None, 'Tamale') # None for crop

for var in pv_variables:
     perform_anova_and_tukey_treatments_retry(tamale_df, var, None, 'Tamale') # None for crop


# Perform analysis for Kumasi
print("\n--- Kumasi Analysis (Retry) ---")
for var in variables_with_crops:
    for crop in crops:
        perform_anova_and_tukey_treatments_retry(kumasi_df, var, crop, 'Kumasi')

for var in ambient_variables:
    perform_anova_and_tukey_treatments_retry(kumasi_df, var, None, 'Kumasi') # None for crop

for var in pv_variables:
     perform_anova_and_tukey_treatments_retry(kumasi_df, var, None, 'Kumasi') # None for crop
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pandas as pd
import numpy as np
import re
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Create directories to save outputs
output_dir_tamale = 'tamale_analysis'
output_dir_kumasi = 'kumasi_analysis'
os.makedirs(output_dir_tamale, exist_ok=True)
os.makedirs(output_dir_kumasi, exist_ok=True)


# Function to perform ANOVA and Tukey's HSD and store results
def perform_anova_and_tukey_store_results(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}")

    cols_to_compare = []
    if crop:
        if 'Soil_' in variable:
             ag_pattern = f'{variable} {crop}'
             av_pattern = f'{variable} AV {crop}'
        elif 'plant canopy' in variable:
             ag_pattern = f'{variable}_AG_{crop}'
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}'
        else:
             ag_pattern = f'{variable}.*AG_{crop}'
             av_pattern = f'{variable}.*AV_{crop}'

        ag_cols = [col for col in df.columns if re.search(ag_pattern, col, re.IGNORECASE)]
        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        cols_to_compare.extend(ag_cols)
        cols_to_compare.extend(av_cols)

    else:
        if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']:
            base_pattern = f'^{variable}$'
            base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
            cols_to_compare.extend(base_cols)
        elif variable in ['PV_Radiation', 'PV_Panel_Temp']:
             base_pattern = f'^{variable}$'
             av_patterns = [f'{variable}_AV_{c}' for c in ['tomato', 'pepper', 'eggplant']] + [f'{variable}_Agrivoltaics plot {c}' for c in ['tomato', 'pepper', 'eggplant']]

             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)


    cols_to_compare = list(set(cols_to_compare))

    if len(cols_to_compare) < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return None, None, None # Return None for tables


    filtered_df = df[cols_to_compare]
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])

    def simplify_treatment_name(col_name):
        if 'AG_' in col_name or ('Soil_Moisture ' in col_name and 'AV' not in col_name) or ('Soil_Temp ' in col_name and 'AV' not in col_name):
            return 'AG'
        elif 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_Panel_Temp' in col_name) and not ('AV' in col_name or 'Agrivoltaics plot' in col_name):
            return 'PV'
        elif '_WS' in col_name or 'Ambient_Humidity' in col_name or 'Ambient_Temp' in col_name or 'Precipitation' in col_name:
            return 'WS'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough unique treatments after dropping NaNs ({data_melted['Treatment'].nunique()}).")
        return None, None, None # Return None for tables

    try:
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)

        if data_melted['Treatment'].nunique() >= 2:
             tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
             tukey_table = pd.DataFrame(data=tukey_result._results_table.data[1:], columns=tukey_result._results_table.data[0])
        else:
             tukey_result = None
             tukey_table = None


        return data_melted, anova_table, tukey_table

    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: {e}")
        return None, None, None # Return None for tables


# Function to create box plots
def create_box_plot(data, variable, crop, location, output_dir, unit=''):
    if data is None or data.empty:
        print(f"Skipping box plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No data available.")
        return

    plt.figure(figsize=(8, 6))
    sns.boxplot(x='Treatment', y='Value', data=data)
    plt.title(f'{variable} ({crop if crop else "Ambient/PV"}) in {location}')
    plt.xlabel('Treatment')
    plt.ylabel(f'{variable}{f" ({unit})" if unit else ""}')
    plt.grid(axis='y', linestyle='--')

    # Save plot
    plot_filename = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_boxplot.png'
    plt.savefig(plot_filename)
    plt.close() # Close the plot to free up memory
    print(f"Saved box plot to {plot_filename}")


# Function to create bar plots with error bars (using descriptive stats)
def create_bar_plot_with_error_bars(data, variable, crop, location, output_dir, unit=''):
     if data is None or data.empty:
        print(f"Skipping bar plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No data available.")
        return

     # Calculate descriptive statistics
     descriptive_stats = data.groupby('Treatment')['Value'].agg(['mean', 'std']).reset_index()

     if descriptive_stats.empty:
         print(f"Skipping bar plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No descriptive statistics available.")
         return

     plt.figure(figsize=(8, 6))
     sns.barplot(x='Treatment', y='mean', yerr=descriptive_stats['std'], data=descriptive_stats, capsize=0.5)
     plt.title(f'Mean {variable} ({crop if crop else "Ambient/PV"}) in {location} with Std Dev Error Bars')
     plt.xlabel('Treatment')
     plt.ylabel(f'Mean {variable}{f" ({unit})" if unit else ""}')
     plt.grid(axis='y', linestyle='--')

     # Save plot
     plot_filename = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_barplot.png'
     plt.savefig(plot_filename)
     plt.close() # Close the plot to free up memory
     print(f"Saved bar plot to {plot_filename}")


# Function to generate descriptive statistics table
def generate_descriptive_stats_table(df, variables, crops, ambient_vars, pv_vars, location, output_dir):
    print(f"\nGenerating descriptive statistics table for {location}")
    all_descriptive_stats = {}

    # Soil and Plant Canopy variables with crops
    for var in variables:
        for crop in crops:
            cols = []
            if 'Soil_' in var:
                cols.extend([f'{var} {crop}', f'{var} AV {crop}'])
            elif 'plant canopy' in var:
                cols.extend([f'{var}_AG_{crop}', f'{var}_Agrivoltaic plot for {crop}'])

            existing_cols = [col for col in cols if col in df.columns]
            if existing_cols:
                temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
                if not temp_df.empty:
                     # Simplify treatment names for the table
                     temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AG' if 'AG' in col_name or ('Soil_' in col_name and 'AV' not in col_name) else 'AV')
                     descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                     all_descriptive_stats[f'{var}_{crop}'] = descriptive_stats


    # Ambient variables
    for var in ambient_vars:
        cols = [var]
        existing_cols = [col for col in cols if col in df.columns]
        if existing_cols:
             temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
             if not temp_df.empty:
                 temp_df['Treatment'] = 'WS' # Assign WS treatment for ambient variables
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    # PV variables
    for var in pv_vars:
         cols = [var] + [f'{var}_AV_{c}' for c in crops] + [f'{var}_Agrivoltaics plot {c}' for c in crops]
         existing_cols = [col for col in cols if col in df.columns]
         if existing_cols:
              temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
              if not temp_df.empty:
                 # Simplify treatment names for PV variables
                 temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AV' if 'AV' in col_name or 'Agrivoltaics plot' in col_name else 'PV')
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    if all_descriptive_stats:
        # Combine all descriptive stats into a single DataFrame for easier saving
        combined_descriptive_stats = pd.concat(all_descriptive_stats, axis=1)
        table_filename = f'{output_dir}/descriptive_statistics.csv'
        combined_descriptive_stats.to_csv(table_filename)
        print(f"Saved descriptive statistics table to {table_filename}")
        display(combined_descriptive_stats.head()) # Display head as table can be large
    else:
        print(f"No descriptive statistics generated for {location}.")


# Store ANOVA and Tukey results
anova_results = {}
tukey_results = {}

# Define variables and crops
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp']


# Perform analysis and generate visualizations/tables for Tamale
print("--- Tamale Analysis Outputs ---")
generate_descriptive_stats_table(tamale_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Tamale', output_dir_tamale)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(tamale_df, var, crop, 'Tamale')
        if data_melted is not None:
            # Create visualizations for AG vs AV
            create_box_plot(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            create_bar_plot_with_error_bars(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Tamale_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Tamale_{var}_{crop}'] = tukey_table

# Perform analysis and generate visualizations/tables for Kumasi
print("\n--- Kumasi Analysis Outputs ---")
generate_descriptive_stats_table(kumasi_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Kumasi', output_dir_kumasi)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(kumasi_df, var, crop, 'Kumasi')
        if data_melted is not None:
             # Create visualizations for AG vs AV
            create_box_plot(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            create_bar_plot_with_error_bars(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Kumasi_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Kumasi_{var}_{crop}'] = tukey_table

# Save ANOVA results to CSV
if anova_results:
    anova_output_filename = 'anova_results.xlsx'
    with pd.ExcelWriter(anova_output_filename) as writer:
        for key, table in anova_results.items():
            table.to_excel(writer, sheet_name=key)
    print(f"\nSaved ANOVA results to {anova_output_filename}")

# Save Tukey HSD results to CSV
if tukey_results:
    tukey_output_filename = 'tukey_hsd_results.xlsx'
    with pd.ExcelWriter(tukey_output_filename) as writer:
        for key, table in tukey_results.items():
             table.to_excel(writer, sheet_name=key)
    print(f"Saved Tukey HSD results to {tukey_output_filename}")
```

The current `descriptive_stats['std'].tolist()` provides a list with the standard deviations for each group, but if there is only one group, this list has a shape of (1,), and if there are two groups, it has a shape of (2,). However, `seaborn.barplot` seems to expect the error bars to be represented as a single value per bar or a (2, n) array where the first row represents the lower bound and the second row represents the upper bound of the error bar.

The error message `ValueError: 'yerr' (shape: (2,)) must be a scalar or a 1D or (2, n) array-like whose shape matches 'y' (shape: (1,))` is misleading when there are two groups. It's likely an internal issue with how `seaborn` handles the `yerr` argument when it's a simple 1D array from a pandas Series or list with exactly two elements.

A more robust approach is to calculate the lower and upper bounds of the error bars manually as `mean - std` and `mean + std` and pass this as a (2, n) array to `yerr`. This explicitly defines the error bar range for each bar.

```python
# Function to create bar plots with error bars (using descriptive stats)
def create_bar_plot_with_error_bars(data, variable, crop, location, output_dir, unit=''):
     if data is None or data.empty:
        print(f"Skipping bar plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No data available.")
        return

     # Calculate descriptive statistics
     descriptive_stats = data.groupby('Treatment')['Value'].agg(['mean', 'std']).reset_index()

     if descriptive_stats.empty:
         print(f"Skipping bar plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No descriptive statistics available.")
         return

     plt.figure(figsize=(8, 6))

     # Define colors
     colors = {'AG': 'green', 'AV': 'blue'}

     # Check if there's more than one group to plot error bars
     if descriptive_stats.shape[0] > 1:
         # Calculate lower and upper bounds for error bars
         yerr_lower = descriptive_stats['mean'] - descriptive_stats['std']
         yerr_upper = descriptive_stats['mean'] + descriptive_stats['std']
         yerr = np.array([yerr_lower, yerr_upper])

         # Use matplotlib for more control over error bars
         x_pos = np.arange(len(descriptive_stats['Treatment']))
         means = descriptive_stats['mean']
         stds = descriptive_stats['std']
         treatments = descriptive_stats['Treatment']

         # Assign colors based on treatment, ensuring the order matches x_pos and means/stds
         bar_colors = [colors.get(t, 'gray') for t in treatments]

         plt.bar(x_pos, means, yerr=stds, align='center', alpha=0.7, capsize=10, color=bar_colors)
         plt.xticks(x_pos, treatments)

         plt.title(f'Mean {variable} ({crop if crop else "Ambient/PV"}) in {location} with Std Dev Error Bars')
         plt.xlabel('Treatment')
         plt.ylabel(f'Mean {variable}{f" ({unit})" if unit else ""}')
         plt.grid(axis='y', linestyle='--')

     else:
         # Use seaborn for a simple bar plot if only one group exists (no error bars needed)
         sns.barplot(x='Treatment', y='mean', data=descriptive_stats, color=colors.get(descriptive_stats['Treatment'].iloc[0], 'gray'))
         plt.title(f'Mean {variable} ({crop if crop else "Ambient/PV"}) in {location}')
         plt.xlabel('Treatment')
         plt.ylabel(f'Mean {variable}{f" ({unit})" if unit else ""}')


     # Remove grid lines
     plt.grid(False)

     # Save plot
     plot_filename = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_barplot.png'
     plt.savefig(plot_filename)
     plt.close() # Close the plot to free up memory
     print(f"Saved bar plot to {plot_filename}")


# Function to perform ANOVA and Tukey's HSD and store results
def perform_anova_and_tukey_store_results(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}")

    cols_to_compare = []
    if crop:
        if 'Soil_' in variable:
             ag_pattern = f'{variable} {crop}'
             av_pattern = f'{variable} AV {crop}'
        elif 'plant canopy' in variable:
             ag_pattern = f'{variable}_AG_{crop}'
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}'
        else:
             ag_pattern = f'{variable}.*AG_{crop}'
             av_pattern = f'{variable}.*AV_{crop}'

        ag_cols = [col for col in df.columns if re.search(ag_pattern, col, re.IGNORECASE)]
        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        cols_to_compare.extend(ag_cols)
        cols_to_compare.extend(av_cols)

    else:
        if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']:
            base_pattern = f'^{variable}$'
            base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
            cols_to_compare.extend(base_cols)
        elif variable in ['PV_Radiation', 'PV_Panel_Temp']:
             base_pattern = f'^{variable}$'
             av_patterns = [f'{variable}_AV_{c}' for c in ['tomato', 'pepper', 'eggplant']] + [f'{variable}_Agrivoltaics plot {c}' for c in ['tomato', 'pepper', 'eggplant']]

             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)


    cols_to_compare = list(set(cols_to_compare))

    if len(cols_to_compare) < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return None, None, None # Return None for tables


    filtered_df = df[cols_to_compare]
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])

    def simplify_treatment_name(col_name):
        if 'AG_' in col_name or ('Soil_Moisture ' in col_name and 'AV' not in col_name) or ('Soil_Temp ' in col_name and 'AV' not in col_name):
            return 'AG'
        elif 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_Panel_Temp' in col_name) and not ('AV' in col_name or 'Agrivoltaics plot' in col_name):
            return 'PV'
        elif '_WS' in col_name or 'Ambient_Humidity' in col_name or 'Ambient_Temp' in col_name or 'Precipitation' in col_name:
            return 'WS'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough unique treatments after dropping NaNs ({data_melted['Treatment'].nunique()}).")
        return None, None, None # Return None for tables

    try:
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)

        if data_melted['Treatment'].nunique() >= 2:
             tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
             tukey_table = pd.DataFrame(data=tukey_result._results_table.data[1:], columns=tukey_result._results_table.data[0])
        else:
             tukey_result = None
             tukey_table = None


        return data_melted, anova_table, tukey_table

    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: {e}")
        return None, None, None # Return None for tables


# Function to create box plots
def create_box_plot(data, variable, crop, location, output_dir, unit=''):
    if data is None or data.empty:
        print(f"Skipping box plot for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: No data available.")
        return

    plt.figure(figsize=(8, 6))
    sns.boxplot(x='Treatment', y='Value', data=data, palette={'AG': 'green', 'AV': 'blue', 'PV': 'gray', 'WS': 'orange'})
    plt.title(f'{variable} ({crop if crop else "Ambient/PV"}) in {location}')
    plt.xlabel('Treatment')
    plt.ylabel(f'{variable}{f" ({unit})" if unit else ""}')
    plt.grid(axis='y', linestyle='--')

    # Save plot
    plot_filename = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_boxplot.png'
    plt.savefig(plot_filename)
    plt.close() # Close the plot to free up memory
    print(f"Saved box plot to {plot_filename}")


# Function to generate descriptive statistics table
def generate_descriptive_stats_table(df, variables, crops, ambient_vars, pv_vars, location, output_dir):
    print(f"\nGenerating descriptive statistics table for {location}")
    all_descriptive_stats = {}

    # Soil and Plant Canopy variables with crops
    for var in variables:
        for crop in crops:
            cols = []
            if 'Soil_' in var:
                cols.extend([f'{var} {crop}', f'{var} AV {crop}'])
            elif 'plant canopy' in var:
                cols.extend([f'{var}_AG_{crop}', f'{var}_Agrivoltaic plot for {crop}'])

            existing_cols = [col for col in cols if col in df.columns]
            if existing_cols:
                temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
                if not temp_df.empty:
                     # Simplify treatment names for the table
                     temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AG' if 'AG' in col_name or ('Soil_' in col_name and 'AV' not in col_name) else 'AV')
                     descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                     all_descriptive_stats[f'{var}_{crop}'] = descriptive_stats


    # Ambient variables
    for var in ambient_vars:
        cols = [var]
        existing_cols = [col for col in cols if col in df.columns]
        if existing_cols:
             temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
             if not temp_df.empty:
                 temp_df['Treatment'] = 'WS' # Assign WS treatment for ambient variables
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    # PV variables
    for var in pv_vars:
         cols = [var] + [f'{var}_AV_{c}' for c in crops] + [f'{var}_Agrivoltaics plot {c}' for c in crops]
         existing_cols = [col for col in cols if col in df.columns]
         if existing_cols:
              temp_df = df[existing_cols].melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])
              if not temp_df.empty:
                 # Simplify treatment names for PV variables
                 temp_df['Treatment'] = temp_df['Treatment'].apply(lambda col_name: 'AV' if 'AV' in col_name or 'Agrivoltaics plot' in col_name else 'PV')
                 descriptive_stats = temp_df.groupby('Treatment')['Value'].agg(['mean', 'median', 'std', 'min', 'max']).T
                 all_descriptive_stats[f'{var}'] = descriptive_stats


    if all_descriptive_stats:
        # Combine all descriptive stats into a single DataFrame for easier saving
        combined_descriptive_stats = pd.concat(all_descriptive_stats, axis=1)
        table_filename = f'{output_dir}/descriptive_statistics.csv'
        combined_descriptive_stats.to_csv(table_filename)
        print(f"Saved descriptive statistics table to {table_filename}")
        display(combined_descriptive_stats.head()) # Display head as table can be large
    else:
        print(f"No descriptive statistics generated for {location}.")


# Store ANOVA and Tukey results
anova_results = {}
tukey_results = {}

# Define variables and crops
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp']


# Perform analysis and generate visualizations/tables for Tamale
print("--- Tamale Analysis Outputs ---")
generate_descriptive_stats_table(tamale_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Tamale', output_dir_tamale)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(tamale_df, var, crop, 'Tamale')
        if data_melted is not None:
            # Create visualizations for AG vs AV
            create_box_plot(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            create_bar_plot_with_error_bars(data_melted, var, crop, 'Tamale', output_dir_tamale, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Tamale_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Tamale_{var}_{crop}'] = tukey_table

# Perform analysis and generate visualizations/tables for Kumasi
print("\n--- Kumasi Analysis Outputs ---")
generate_descriptive_stats_table(kumasi_df, variables_with_crops, crops, ambient_variables, pv_variables, 'Kumasi', output_dir_kumasi)

for var in variables_with_crops:
    for crop in crops:
        data_melted, anova_table, tukey_table = perform_anova_and_tukey_store_results(kumasi_df, var, crop, 'Kumasi')
        if data_melted is not None:
             # Create visualizations for AG vs AV
            create_box_plot(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')
            create_bar_plot_with_error_bars(data_melted, var, crop, 'Kumasi', output_dir_kumasi, unit='%' if 'Humidity' in var or 'Moisture' in var else 'oC' if 'Temp' in var else '')

            # Store ANOVA and Tukey results
            if anova_table is not None:
                anova_results[f'Kumasi_{var}_{crop}'] = anova_table
            if tukey_table is not None:
                tukey_results[f'Kumasi_{var}_{crop}'] = tukey_table

# Save ANOVA results to Excel
if anova_results:
    anova_output_filename = 'anova_results.xlsx'
    with pd.ExcelWriter(anova_output_filename) as writer:
        for key, table in anova_results.items():
            table.to_excel(writer, sheet_name=key)
    print(f"\nSaved ANOVA results to {anova_output_filename}")

# Save Tukey HSD results to Excel
if tukey_results:
    tukey_output_filename = 'tukey_hsd_results.xlsx'
    with pd.ExcelWriter(tukey_output_filename) as writer:
        for key, table in tukey_results.items():
             table.to_excel(writer, sheet_name=key)
    print(f"Saved Tukey HSD results to {tukey_output_filename}")
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import numpy as np
import re

# Function to perform ANOVA and Tukey's HSD post-hoc test for specific treatments
def perform_anova_and_tukey_treatments_retry(df, variable, crop, location):
    print(f"\nPerforming ANOVA and Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}")

    # Dynamically find columns based on variable and crop patterns
    cols_to_compare = []
    if crop:
        # For variables with crops, look for AG_ and AV_ treatments
        # Handle Soil variables
        if 'Soil_' in variable:
             ag_pattern = f'{variable} {crop}'
             av_pattern = f'{variable} AV {crop}'
        # Handle Plant canopy variables
        elif 'plant canopy' in variable:
             ag_pattern = f'{variable}_AG_{crop}'
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}'
        else:
             ag_pattern = f'{variable}.*AG_{crop}'
             av_pattern = f'{variable}.*AV_{crop}'


        # Find columns matching the patterns
        ag_cols = [col for col in df.columns if re.search(ag_pattern, col, re.IGNORECASE)]
        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        cols_to_compare.extend(ag_cols)
        cols_to_compare.extend(av_cols)

    else:
        # For ambient/PV variables, look for relevant columns.
        if variable in ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']:
            # These variables appear to have only one column representing the WS treatment, so skip ANOVA
            print(f"Skipping {variable} in {location}: Only one treatment (WS) available.")
            return
        elif variable in ['PV_Radiation', 'PV_Panel_Temp']:
             # Look for base PV, and AV specific PV columns for each crop
             base_pattern = f'^{variable}$'
             # Collect all AV-related PV columns regardless of crop when crop is None
             av_patterns = [f'{variable}_AV_{c}' for c in crops] + [f'{variable}_Agrivoltaics plot {c}' for c in crops]

             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)


    # Ensure unique columns in case of overlapping patterns
    cols_to_compare = list(set(cols_to_compare))


    if len(cols_to_compare) < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return

    # Filter the DataFrame to include only the identified column names
    filtered_df = df[cols_to_compare]

    # Melt the filtered DataFrame for ANOVA
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value')

    # Extract simplified treatment names ('AG', 'AV', 'PV', 'WS')
    def simplify_treatment_name(col_name):
        if 'AG_' in col_name or ('Soil_Moisture ' in col_name and 'AV' not in col_name) or ('Soil_Temp ' in col_name and 'AV' not in col_name):
            return 'AG'
        elif 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_Panel_Temp' in col_name) and not ('AV' in col_name or 'Agrivoltaics plot' in col_name):
            # This condition tries to capture the base PV columns as 'PV'
            return 'PV'
        elif '_WS' in col_name or 'Ambient_Humidity' in col_name or 'Ambient_Temp' in col_name or 'Precipitation' in col_name:
            return 'WS'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)


    # Drop any rows with NaN values in the 'Value' column
    data_melted = data_melted.dropna(subset=['Value'])

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping {variable}" + (f" ({crop})" if crop else "") + f" in {location}: Not enough unique treatments after dropping NaNs ({data_melted['Treatment'].nunique()}).")
        return

    try:
        # Perform ANOVA
        print("ANOVA Results:")
        model = ols('Value ~ C(Treatment)', data=data_melted).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        display(anova_table)

        # Perform Tukey's HSD post-hoc test
        if data_melted['Treatment'].nunique() >= 2:
            print("\nTukey's HSD Post-Hoc Test Results:")
            tukey_result = pairwise_tukeyhsd(endog=data_melted['Value'],
                                           groups=data_melted['Treatment'],
                                           alpha=0.05)
            display(tukey_result)
        else:
            print("\nSkipping Tukey's HSD: Not enough unique groups for comparison after dropping NaNs.")

    except Exception as e:
        print(f"An error occurred during ANOVA or Tukey's HSD for {variable}" + (f" ({crop})" if crop else "") + f" in {location}: {e}")

# Define variables and crops based on observed column names
# Note the long names for ambient/canopy variables
variables_with_crops = ['Soil_Moisture', 'Soil_Temp', 'Relative humidity of air with plant canopy', 'Air temperature within plant canopy']
crops = ['tomato', 'pepper', 'eggplant']
ambient_variables = ['Ambient_Humidity', 'Ambient_Temp', 'Precipitation']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp']

locations = {'Tamale': tamale_df, 'Kumasi': kumasi_df}

# Perform analysis for Tamale
print("--- Tamale Analysis (Retry) ---")
for var in variables_with_crops:
    for crop in crops:
        perform_anova_and_tukey_treatments_retry(locations['Tamale'], var, crop, 'Tamale')

for var in ambient_variables:
    perform_anova_and_tukey_treatments_retry(locations['Tamale'], var, None, 'Tamale') # None for crop

for var in pv_variables:
     perform_anova_and_tukey_treatments_retry(locations['Tamale'], var, None, 'Tamale') # None for crop


# Perform analysis for Kumasi
print("\n--- Kumasi Analysis (Retry) ---")
for var in variables_with_crops:
    for crop in crops:
        perform_anova_and_tukey_treatments_retry(locations['Kumasi'], var, crop, 'Kumasi')

for var in ambient_variables:
    perform_anova_and_tukey_treatments_retry(locations['Kumasi'], var, None, 'Kumasi') # None for crop

for var in pv_variables:
     perform_anova_and_tukey_treatments_retry(locations['Kumasi'], var, None, 'Kumasi') # None for crop
```

```python
import plotly.express as px
import pandas as pd
import re
import os

# Create directories to save outputs
output_dir_tamale = 'tamale_analysis'
output_dir_kumasi = 'kumasi_analysis'
os.makedirs(output_dir_tamale, exist_ok=True)
os.makedirs(output_dir_kumasi, exist_ok=True)

# Function to create box plots using Plotly Express for AV vs PV comparison
def create_plotly_box_plot_av_pv(df, variable, crop, location, output_dir, unit=''):
    print(f"\nCreating Plotly box plot for {variable}" + (f" ({crop})" if crop else "") + f" (AV vs PV) in {location}")

    cols_to_compare = []
    # For variables with crops, look for AV_ treatments and the base PV column
    if crop:
        # Construct AV column patterns for PV variables with crops
        if 'PV_Radiation' in variable:
             av_pattern = f'PV_Radiation_Agrivoltaics plot {crop}' if crop == 'tomato' else f'PV_Radiation_AV_{crop}'
        elif 'PV_PanelTemp' in variable:
             av_pattern = f'PV_PanelTemp AV_{crop}'
        else: # For non-PV variables with crops, look for the standard AV pattern
             av_pattern = f'{variable}_Agrivoltaic plot for {crop}' if 'plant canopy' in variable else f'{variable} AV {crop}'


        av_cols = [col for col in df.columns if re.search(av_pattern, col, re.IGNORECASE)]

        # Add the base PV column if it exists
        base_pv_col = variable.replace('_AV', '').replace('_Agrivoltaics plot', '').strip() # Get base name
        if base_pv_col in df.columns:
             cols_to_compare.append(base_pv_col)
        cols_to_compare.extend(av_cols)

    else:
        # For PV variables without specific crop treatments, compare base PV with all AV PV columns
        if variable in ['PV_Radiation', 'PV_PanelTemp']:
             base_pattern = f'^{variable}$'
             # Collect all AV-related PV columns across all crops
             av_patterns = [f'PV_Radiation_Agrivoltaics plot {c}' for c in ['tomato', 'pepper', 'eggplant']] + \
                           [f'PV_Radiation_AV_{c}' for c in ['pepper', 'eggplant']] + \
                           [f'PV_PanelTemp AV_{c}' for c in ['tomato', 'pepper', 'eggplant']]


             base_cols = [col for col in df.columns if re.search(base_pattern, col, re.IGNORECASE)]
             av_cols = [col for col in df.columns for pattern in av_patterns if re.search(pattern, col, re.IGNORECASE)]

             cols_to_compare.extend(base_cols)
             cols_to_compare.extend(av_cols)
        else:
            print(f"Skipping plotting for {variable}" + (f" ({crop})" if crop else "") + f" (AV vs PV) in {location}: Variable not relevant for AV vs PV comparison.")
            return


    cols_to_compare = list(set(cols_to_compare))

    if len(cols_to_compare) < 2:
        print(f"Skipping Plotly box plot for {variable}" + (f" ({crop})" if crop else "") + f" (AV vs PV) in {location}: Not enough relevant columns ({len(cols_to_compare)}).")
        return

    filtered_df = df[cols_to_compare]
    data_melted = filtered_df.melt(var_name='Treatment', value_name='Value').dropna(subset=['Value'])

    # Extract simplified treatment names ('AV', 'PV')
    def simplify_treatment_name(col_name):
        if 'AV_' in col_name or 'Agrivoltaic plot' in col_name or ('Soil_Moisture AV ' in col_name) or ('Soil_Temp AV ' in col_name):
            return 'AV'
        elif ('_PV' in col_name or 'PV_Radiation' in col_name or 'PV_PanelTemp' in col_name) and not re.search(r'AV_|Agrivoltaic plot', col_name, re.IGNORECASE):
             # This condition tries to capture the base PV columns as 'PV', excluding AV-related ones
            return 'PV'
        else:
            return 'Other'

    data_melted['Treatment'] = data_melted['Treatment'].apply(simplify_treatment_name)

    # Filter for only AV and PV treatments
    data_melted = data_melted[data_melted['Treatment'].isin(['AV', 'PV'])]

    if data_melted['Treatment'].nunique() < 2:
        print(f"Skipping Plotly box plot for {variable}" + (f" ({crop})" if crop else "") + f" (AV vs PV) in {location}: Not enough unique treatments after filtering ({data_melted['Treatment'].nunique()}).")
        return

    fig = px.box(data_melted,
                 x="Treatment",
                 y="Value",
                 points="all",
                 title=f'{variable} ({crop if crop else "PV"}) in {location} by Treatment (AV vs PV)',
                 labels={'Treatment': 'Treatment', 'Value': f'{variable}{f" ({unit})" if unit else ""}'},
                 color="Treatment", # Color by treatment
                 color_discrete_map={'AV': 'green', 'PV': 'blue'}) # Map AV to green and PV to blue


    # Save plot as HTML
    plot_filename_html = f'{output_dir}/{variable.replace(" ", "_")}{"_" + crop if crop else ""}_AV_PV_boxplot.html'
    fig.write_html(plot_filename_html)
    print(f"Saved Plotly box plot to {plot_filename_html}")


# Define variables and crops
crops = ['tomato', 'pepper', 'eggplant']
pv_variables = ['PV_Radiation', 'PV_Panel_Temp'] # Variables to compare AV vs PV without crop specificity

# Define locations
locations = {'Tamale': tamale_df, 'Kumasi': kumasi_df}


# Perform analysis and generate visualizations for Tamale (AV vs PV)
print("--- Tamale Analysis (AV vs PV Plots) ---")
# For PV variables, compare base PV with all AV PV columns (no crop specified)
for var in pv_variables:
     create_plotly_box_plot_av_pv(locations['Tamale'], var, None, 'Tamale', output_dir_tamale, unit='W/m²' if 'Radiation' in var else 'oC' if 'Temp' in var else '')

# Perform analysis and generate visualizations for Kumasi (AV vs PV)
print("\n--- Kumasi Analysis (AV vs PV Plots) ---")
# For PV variables, compare base PV with all AV PV columns (no crop specified)
for var in pv_variables:
     create_plotly_box_plot_av_pv(locations['Kumasi'], var, None, 'Kumasi', output_dir_kumasi, unit='W/m²' if 'Radiation' in var else 'oC' if 'Temp' in var else '')
```
