## Implement analysis based on prompt

### Subtask:
Analyze the microclimate data for Kumasi and Tamale, focusing on the comparisons between AV and AG treatments for each crop (tomato, pepper, eggplant) and the ambient weather station data. Calculate and compare daily mean soil moisture, soil temperature, canopy relative humidity, and canopy temperature for each treatment and crop at both locations.


**Reasoning**:
Calculate and compare daily mean soil moisture, soil temperature, canopy relative humidity, and canopy temperature for each treatment and crop at both locations using the daily dataframes.



```python
import pandas as pd

locations = {'Kumasi': df_kumasi_daily, 'Tamale': df_tamale_daily}
crops = ['tomato', 'pepper', 'eggplant']
parameters = ['soil_moisture', 'soil_temp', 'canopy_rh', 'canopy_temp']

analysis_results = {}

for location_name, df_daily in locations.items():
    analysis_results[location_name] = {}
    for crop in crops:
        analysis_results[location_name][crop] = {}
        for param in parameters:
            ag_col = f'{param}_ag_{crop}'
            av_col = f'{param}_av_{crop}'

            if ag_col in df_daily.columns and av_col in df_daily.columns:
                # Calculate daily means (already done in the previous step by resampling)
                # Calculate difference and ratio
                difference = df_daily[av_col] - df_daily[ag_col]
                ratio = df_daily[av_col] / df_daily[ag_col]

                analysis_results[location_name][crop][param] = {
                    'ag_daily_mean': df_daily[ag_col],
                    'av_daily_mean': df_daily[av_col],
                    'difference': difference,
                    'ratio': ratio
                }
            else:
                print(f"Warning: Columns for {param} of {crop} at {location_name} not found.")

# Display some results, e.g., daily mean soil moisture difference for tomato in Kumasi
display("Kumasi Tomato Soil Moisture Daily Difference (AV - AG):")
display(analysis_results['Kumasi']['tomato']['soil_moisture']['difference'].head())

display("Tamale Pepper Canopy Temperature Daily Ratio (AV / AG):")
display(analysis_results['Tamale']['pepper']['canopy_temp']['ratio'].head())
```

## Visualize results

Visualize the analysis results to highlight the microclimate differences between AV and AG treatments for each crop at both locations. **Reasoning**:
Visualize the analysis results by iterating through the analysis_results dictionary and creating plots for daily means and differences/ratios for each parameter, crop, and location.



```python
import matplotlib.pyplot as plt

locations = ['Kumasi', 'Tamale']
crops = ['tomato', 'pepper', 'eggplant']
parameters = ['soil_moisture', 'soil_temp', 'canopy_rh', 'canopy_temp']
param_units = {
    'soil_moisture': '%',
    'soil_temp': '°C',
    'canopy_rh': '%',
    'canopy_temp': '°C'
}
# Determine which metric to plot for difference/ratio
comparison_metric = {
    'soil_moisture': 'difference',
    'soil_temp': 'difference',
    'canopy_rh': 'difference',
    'canopy_temp': 'difference'
}


for location_name in locations:
    for crop in crops:
        fig, axes = plt.subplots(nrows=len(parameters), ncols=2, figsize=(15, len(parameters) * 5))
        fig.suptitle(f'{location_name} - {crop} Microclimate Analysis', y=1.02, fontsize=16)

        for i, param in enumerate(parameters):
            if param in analysis_results[location_name][crop]:
                # Plot daily means
                analysis_results[location_name][crop][param]['ag_daily_mean'].plot(ax=axes[i, 0], label='AG', color='blue')
                analysis_results[location_name][crop][param]['av_daily_mean'].plot(ax=axes[i, 0], label='AV', color='red')
                axes[i, 0].set_title(f'{param.replace("_", " ").title()}: Daily Means')
                axes[i, 0].set_ylabel(f'{param.replace("_", " ").title()} ({param_units[param]})')
                axes[i, 0].legend()
                axes[i, 0].grid(True)

                # Plot difference or ratio
                metric_to_plot = comparison_metric[param]
                analysis_results[location_name][crop][param][metric_to_plot].plot(ax=axes[i, 1], color='purple')
                axes[i, 1].set_title(f'{param.replace("_", " ").title()}: Daily {metric_to_plot.title()} (AV - AG)')
                axes[i, 1].set_ylabel(f'{param.replace("_", " ").title()} {metric_to_plot.title()} ({param_units[param]})')
                axes[i, 1].grid(True)
            else:
                axes[i, 0].set_title(f'{param.replace("_", " ").title()}: Data Not Available')
                axes[i, 1].set_title(f'{param.replace("_", " ").title()}: Data Not Available')
                axes[i, 0].axis('off')
                axes[i, 1].axis('off')


        plt.tight_layout()
        plt.show()
```

## Visualize Treatment Comparisons with Plotly Box Plots

### Subtask:
Generate box plots using Plotly to visualize the differences in microclimate parameters between treatments (AG vs AV, and AV vs PV) for each crop at both locations. Use the specified color scheme (AG: green, AV: blue, PV: gray).

**Reasoning**:
Generate box plots using Plotly to visually compare the distributions of microclimate parameters across different treatments (AG vs AV, and AV vs PV) for each crop and location. This aligns with the prompt's requirement for advanced visualizations for academic papers and thesis chapters. The specified color scheme will be used for clarity.

```python
import plotly.express as px
import os

# Create a combined dataframe with a 'Treatment' column for easier plotting
# We'll focus on the daily data since the previous analysis and prompt emphasize daily patterns

# First, unpivot the daily dataframes to long format
df_kumasi_daily_long = df_kumasi_daily.reset_index().melt(id_vars='DateTime', var_name='Variable')
df_tamale_daily_long = df_tamale_daily.reset_index().melt(id_vars='DateTime', var_name='Variable')

# Add 'Location' and 'Treatment' columns
df_kumasi_daily_long['Location'] = 'Kumasi'
df_tamale_daily_long['Location'] = 'Tamale'

# Combine into a single dataframe
df_combined_daily = pd.concat([df_kumasi_daily_long, df_tamale_daily_long])

# Extract 'Treatment', 'Crop', and 'Parameter' from the 'Variable' column
def extract_info(variable):
    parts = variable.split('_')
    if parts[-2] == 'ag':
        treatment = 'AG'
    elif parts[-2] == 'av':
        treatment = 'AV'
    elif parts[-2] == 'pv':
        treatment = 'PV'
    elif parts[-1] == 'ws':
        treatment = 'WS'
    else:
        treatment = parts[-2] # Keep original if not one of the expected treatments

    if parts[-1] in ['tomato', 'pepper', 'eggplant']:
        crop = parts[-1]
        param = '_'.join(parts[:-2])
    elif parts[-1] == 'ws':
        crop = 'N/A' # Weather station not crop-specific
        param = '_'.join(parts[:-1])
    elif parts[-2] == 'only':
        crop = 'only' # PV only treatment
        param = '_'.join(parts[:-2])
    else:
      # Handle cases where variable name doesn't match expected pattern
        param = '_'.join(parts)
        treatment = 'Unknown'
        crop = 'Unknown'


    return param, treatment, crop

df_combined_daily[['Parameter', 'Treatment', 'Crop']] = df_combined_daily['Variable'].apply(lambda x: pd.Series(extract_info(x)))

# Define parameters and their display names and units
parameters_to_plot = {
    'soil_moisture': {'name': 'Soil Moisture', 'unit': '%'},
    'soil_temp': {'name': 'Soil Temperature', 'unit': '°C'},
    'canopy_rh': {'name': 'Canopy Relative Humidity', 'unit': '%'},
    'canopy_temp': {'name': 'Canopy Temperature', 'unit': '°C'},
    'pv_solar_radiation': {'name': 'PV Solar Radiation', 'unit': 'W/m²'},
    'pv_undestory_panel_temperature': {'name': 'PV Undestory Panel Temperature', 'unit': '°C'}

}

# Define colors for treatments
color_map = {'AG': 'green', 'AV': 'blue', 'PV': 'gray'}

# Create a directory to save the plots
output_dir = "Results_Figures"
os.makedirs(output_dir, exist_ok=True)


# Generate box plots for AG vs AV for each location, crop, and parameter
for location in df_combined_daily['Location'].unique():
    for crop in ['tomato', 'pepper', 'eggplant']: # Iterate through crops for AV and AG
        for parameter, info in parameters_to_plot.items():
            df_plot = df_combined_daily[(df_combined_daily['Location'] == location) &
                                        (df_combined_daily['Crop'] == crop) &
                                        (df_combined_daily['Parameter'] == parameter) &
                                        (df_combined_daily['Treatment'].isin(['AG', 'AV']))]


            if not df_plot.empty:
                fig = px.box(df_plot,
                             x='Treatment',
                             y='value',
                             color='Treatment',
                             color_discrete_map=color_map,
                             title=f'{info["name"]} Comparison for {crop.title()} in {location}',
                             labels={'value': f'{info["name"]} ({info["unit"]})', 'Treatment': 'Treatment'},
                             points="all") # Added points="all"
                fig.update_layout(yaxis_title=f'{info["name"]} ({info["unit"]})')
                fig.show()
                # Save the plot
                filename = f"{location}_{crop}_{parameter}_AG_vs_AV_boxplot.html"
                filepath = os.path.join(output_dir, filename)
                fig.write_html(filepath)


# Generate box plots for AV vs PV for each location and parameter (PV is not crop specific, so compare with AV for each crop)
for location in df_combined_daily['Location'].unique():
    for crop in ['tomato', 'pepper', 'eggplant']: # Iterate through crops for AV
        for parameter, info in parameters_to_plot.items():
             # Filter for AV data for the specific crop and PV data (crop='only')
            df_plot = df_combined_daily[(df_combined_daily['Location'] == location) &
                                        (df_combined_daily['Parameter'] == parameter) &
                                        ((df_combined_daily['Treatment'] == 'AV') & (df_combined_daily['Crop'] == crop) |
                                         (df_combined_daily['Treatment'] == 'PV') & (df_combined_daily['Crop'] == 'only'))
                                        ]


            if not df_plot.empty:
                fig = px.box(df_plot,
                             x='Treatment',
                             y='value',
                             color='Treatment',
                             color_discrete_map=color_map,
                             title=f'{info["name"]} Comparison (AV vs PV for {crop.title()}) in {location}',
                             labels={'value': f'{info["name"]} ({info["unit"]})', 'Treatment': 'Treatment'},
                             points="all") # Added points="all"
                fig.update_layout(yaxis_title=f'{info["name"]} ({info["unit"]})')
                fig.show()
                # Save the plot
                filename = f"{location}_{crop}_{parameter}_AV_vs_PV_boxplot.html"
                filepath = os.path.join(output_dir, filename)
                fig.write_html(filepath)
```

## Visualize PV vs AV Comparison with Plotly Box Plots

### Subtask:
Generate box plots using Plotly to visualize the differences in PV system parameters (PV solar radiation and PV undestory panel temperature) between the Agrivoltaics (AV) and Photovoltaics Only (PV) treatments at both locations. Use the specified color scheme (AV: blue, PV: gray).

**Reasoning**:
To compare the performance of the PV system under agrivoltaic conditions (AV) with the conventional photovoltaic setup (PV), generate box plots for PV solar radiation and PV undestory panel temperature. This visualization will help assess how the presence of crops and altered microclimate in the AV system affects PV performance compared to the PV-only setup.

```python
import plotly.express as px
import os

# Create a combined dataframe with a 'Treatment' column for easier plotting
# We'll focus on the daily data

# First, unpivot the daily dataframes to long format
df_kumasi_daily_long = df_kumasi_daily.reset_index().melt(id_vars='DateTime', var_name='Variable')
df_tamale_daily_long = df_tamale_daily.reset_index().melt(id_vars='DateTime', var_name='Variable')

# Add 'Location' and 'Treatment' columns
df_kumasi_daily_long['Location'] = 'Kumasi'
df_tamale_daily_long['Location'] = 'Tamale'

# Combine into a single dataframe
df_combined_daily = pd.concat([df_kumasi_daily_long, df_tamale_daily_long])

# Extract 'Treatment', 'Crop', and 'Parameter' from the 'Variable' column
def extract_info(variable):
    parts = variable.split('_')
    if parts[-2] == 'ag':
        treatment = 'AG'
    elif parts[-2] == 'av':
        treatment = 'AV'
    elif parts[-2] == 'pv':
        treatment = 'PV'
    elif parts[-1] == 'ws':
        treatment = 'WS'
    else:
        treatment = parts[-2] # Keep original if not one of the expected treatments

    if parts[-1] in ['tomato', 'pepper', 'eggplant']:
        crop = parts[-1]
        param = '_'.join(parts[:-2])
    elif parts[-1] == 'ws':
        crop = 'N/A' # Weather station not crop-specific
        param = '_'.join(parts[:-1])
    elif parts[-2] == 'only':
        crop = 'only' # PV only treatment
        param = '_'.join(parts[:-2])
    else:
      # Handle cases where variable name doesn't match expected pattern
        param = '_'.join(parts)
        treatment = 'Unknown'
        crop = 'Unknown'


    return param, treatment, crop


df_combined_daily[['Parameter', 'Treatment', 'Crop']] = df_combined_daily['Variable'].apply(lambda x: pd.Series(extract_info(x)))

# Define PV parameters and their display names and units
pv_parameters_to_plot = {
    'pv_solar_radiation': {'name': 'PV Solar Radiation', 'unit': 'W/m²'},
    'pv_undestory_panel_temperature': {'name': 'PV Undestory Panel Temperature', 'unit': '°C'}
}

# Define colors for treatments
color_map = {'AG': 'green', 'AV': 'blue', 'PV': 'gray'}

# Create a directory to save the plots
output_dir = "Results_Figures"
os.makedirs(output_dir, exist_ok=True)


# Generate box plots for AV vs PV for each location and PV parameter
for location in df_combined_daily['Location'].unique():
    for parameter, info in pv_parameters_to_plot.items():
         # Filter for AV data for all crops and PV data (crop='only')
        df_plot = df_combined_daily[(df_combined_daily['Location'] == location) &
                                    (df_combined_daily['Parameter'] == parameter) &
                                    ((df_combined_daily['Treatment'] == 'AV') |
                                     (df_combined_daily['Treatment'] == 'PV') & (df_combined_daily['Crop'] == 'only'))
                                    ]

        if not df_plot.empty:
            fig = px.box(df_plot,
                         x='Treatment',
                         y='value',
                         color='Treatment',
                         color_discrete_map=color_map,
                         title=f'{info["name"]} Comparison (AV vs PV) in {location}',
                         labels={'value': f'{info["name"]} ({info["unit"]})', 'Treatment': 'Treatment'},
                         points="all") # Added points="all"
            fig.update_layout(yaxis_title=f'{info["name"]} ({info["unit"]})')
            fig.show()
            # Save the plot
            filename = f"{location}_{parameter}_AV_vs_PV_boxplot.html"
            filepath = os.path.join(output_dir, filename)
            fig.write_html(filepath)
```
